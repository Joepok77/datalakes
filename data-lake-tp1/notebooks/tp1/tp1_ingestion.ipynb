```python
import boto3

s3 = boto3.client(
    "s3",
    endpoint_url="http://minio:9000",
    aws_access_key_id="minioadmin",
    aws_secret_access_key="minioadmin123",
)

# Ingestion du CSV dans la zone raw, partitionnee par date
s3.upload_file("ventes.csv", "raw", "ventes/2026/09/11/ventes.csv")

# Ingestion du JSON dans la zone raw
s3.upload_file("clients.json", "raw", "clients/2026/09/11/clients.json")

# Verification : lister le contenu de la zone raw
response = s3.list_objects_v2(Bucket="raw")
for obj in response.get("Contents", []):
    print(obj["Key"], "-", obj["Size"], "octets")
```
